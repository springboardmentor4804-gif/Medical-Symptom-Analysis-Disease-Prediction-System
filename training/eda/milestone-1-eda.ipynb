```python

def profile_df(df, name):
    print(f"===== {name} =====")
    print("Shape:", df.shape)
    print("\nDtypes:\n", df.dtypes)
    print("\nMissing values:\n", df.isnull().sum())
    print("\nDuplicate rows (exact):", df.duplicated().sum())
    print()

profile_df(df1, "Dataset 1 - Patient Profile")
profile_df(df2, "Dataset 2 - Symptoms/Cures")

# Standardize column names for both datasets
df1.columns = df1.columns.str.strip().str.lower().str.replace(" ", "_")
df2.columns = df2.columns.str.strip().str.lower().str.replace(" ", "_")

# Drop exact duplicate rows (49 in df1, 3 in df2 in the original exploration)
df1 = df1.drop_duplicates().reset_index(drop=True)
df2 = df2.drop_duplicates().reset_index(drop=True)
print(f"\nAfter exact-duplicate removal -> df1: {df1.shape}, df2: {df2.shape}")
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.3</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Disease-name normalization &amp; deduplication</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Diseases are often written differently across the two files (like "flu" vs "influenza", or "UTI" vs "urinary tract infection"). This step cleans up and matches those names so the same disease is recognized as one thing, not several.</p><p>Disease names differ in case, punctuation, parenthetical abbreviations (<code>"chronic obstructive pulmonary (COPD)"</code> vs <code>"chronic obstructive pulmonary"</code>), and true synonyms (<code>"flu"</code> vs <code>"influenza"</code>) that simple string similarity cannot catch. This step runs in three passes:</p><ol style="padding-left:20px;margin:10px 0;"><li style="margin-bottom:5px;"><b>Normalize</b> — lowercase, strip punctuation/parentheses, drop the trailing word "disease".</li><li style="margin-bottom:5px;"><b>Self-dedup within each dataset</b> — catches cases where the <i>same</i> disease was written two different ways in the <i>same</i> dataset (e.g. "urinary tract infection" vs "urinary tract infection uti").</li><li style="margin-bottom:5px;"><b>Cross-dataset alias mapping</b> — fuzzy string matching for spelling variants, plus a hand-built synonym dictionary for true synonyms that character-similarity cannot catch.</li></ol></div></div>

```python

# ============================================================
# STEP 2 — Normalize disease names
# ============================================================
def normalize_name(name):
    if pd.isna(name):
        return name
    n = str(name).lower().strip()
    n = re.sub(r"\(.*?\)", "", n)              # strip parenthetical abbreviations, e.g. "(COPD)"
    n = re.sub(r"[^a-z0-9\s]", "", n)          # strip punctuation/apostrophes
    n = re.sub(r"\s+", " ", n)
    n = re.sub(r"\bdisease\b", "", n).strip()  # drop trailing "disease"
    n = re.sub(r"\s+", " ", n).strip()
    return n

df1["disease_clean"] = df1["disease"].apply(normalize_name)
df2["disease_clean"] = df2["disease"].apply(normalize_name)

print("Unique diseases after normalization -> df1:", df1["disease_clean"].nunique(),
      "| df2:", df2["disease_clean"].nunique())
```

```python

# ============================================================
# STEP 3 — Self-dedup WITHIN each dataset
# Catches "X" vs "X uti" / "X copd" style splits of the SAME disease
# ============================================================
def self_dedup_map(names_unique):
    alias = {}
    for n in names_unique:
        for other in names_unique:
            if other == n:
                continue
            n_words, other_words = set(n.split()), set(other.split())
            if n_words.issubset(other_words) and len(n_words) >= 2:
                longer, shorter = (other, n) if len(other) > len(n) else (n, other)
                alias[longer] = shorter
    return alias

self_alias_df1 = self_dedup_map(sorted(df1["disease_clean"].dropna().unique()))
self_alias_df2 = self_dedup_map(sorted(df2["disease_clean"].dropna().unique()))

df1["disease_clean"] = df1["disease_clean"].replace(self_alias_df1)
df2["disease_clean"] = df2["disease_clean"].replace(self_alias_df2)

print(f"Self-dedup pairs merged within df1: {len(self_alias_df1)}")
print(f"Self-dedup pairs merged within df2: {len(self_alias_df2)}")
print(f"Unique diseases after self-dedup -> df1: {df1['disease_clean'].nunique()}, df2: {df2['disease_clean'].nunique()}")
```

```python

# ============================================================
# STEP 4 — Cross-dataset alias mapping (df2 name -> df1 canonical name)
# Fuzzy matches spelling variants; manual dict handles true synonyms
# ============================================================
from difflib import get_close_matches

def best_match(name, choices, cutoff=0.82):
    matches = get_close_matches(name, choices, n=1, cutoff=cutoff)
    return matches[0] if matches else None

names1 = sorted(df1["disease_clean"].dropna().unique())
names2 = sorted(df2["disease_clean"].dropna().unique())

alias_map = {}
for n2 in names2:
    m = best_match(n2, names1)
    if m and m != n2:
        alias_map[n2] = m

# True synonyms fuzzy matching cannot catch (identified during exploratory analysis)
manual_alias_overrides = {
    "heart attack": "myocardial infarction heart",
    "conjunctivitis": "conjunctivitis pink eye",
    "eating disorder": "eating disorders anorexia",
    "lupus": "systemic lupus erythematosus",
    # NOTE: "arthritis" -> "osteoarthritis" was tested and rejected: it conflates a
    # general condition (df2's "arthritis") with one specific subtype (df1's
    # "osteoarthritis"). Left out deliberately to avoid a semantically fuzzy merge.
}
alias_map.update(manual_alias_overrides)

df2["disease_canonical"] = df2["disease_clean"].replace(alias_map)
df1["disease_canonical"] = df1["disease_clean"]

overlap = set(df1["disease_canonical"]) & set(df2["disease_canonical"])
print(f"Diseases in df1 only: {len(set(df1['disease_canonical']) - overlap)}")
print(f"Diseases in df2 only: {len(set(df2['disease_canonical']) - overlap)}")
print(f"Overlapping diseases: {len(overlap)}")
print(f"\nCross-dataset aliases applied ({len(alias_map)}):")
for k, v in alias_map.items():
    print(f"  '{k}'  ->  '{v}'")
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.4</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Symptom encoding, categorical cleanup &amp; risk-level parsing</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> We turn "Yes"/"No" answers into the numbers 1 and 0 that a computer can work with, tidy up messy text categories, and fix a badly-formatted risk field so it's actually usable.</p><p>Yes/No symptom flags are converted to binary, categorical text is standardized (blank strings treated as genuinely missing, not a category), and the malformed <code>risk_level</code> field is parsed into a clean ordinal category plus a numeric percentage where one is stated.</p></div></div>

```python

# ============================================================
# STEP 5 — Convert Yes/No symptom flags to binary, clean categoricals,
#          parse the malformed risk_level field into risk_category + risk_pct
# ============================================================
symptom_cols = [c for c in ["fever", "cough", "fatigue", "difficulty_breathing"] if c in df1.columns]
for c in symptom_cols:
    df1[c] = df1[c].astype(str).str.strip().str.lower().map({"yes": 1, "no": 0})

def clean_categorical(series):
    s = series.astype(str).str.strip().str.lower()
    return s.replace({"nan": np.nan, "": np.nan})   # blank strings ARE missing, not a real category

cat_cols_df1 = ["gender", "blood_pressure", "cholesterol_level", "outcome_variable"]
for c in cat_cols_df1:
    df1[c] = clean_categorical(df1[c])
for c in ["cures", "doctor", "symptoms"]:
    df2[c] = clean_categorical(df2[c])

# risk_level is malformed in the raw data: "low (0.1%" (missing closing paren),
# "varies depending on the type of cancer", etc. Extract a clean ordinal category
# plus a numeric percentage where one is stated.
def clean_risk_level(val):
    if pd.isna(val):
        return np.nan, np.nan
    v = str(val).strip().lower()
    pct_match = re.search(r"(\d+(?:\.\d+)?)\s*%", v)
    risk_pct = float(pct_match.group(1)) if pct_match else np.nan
    if "varies" in v: risk_category = "varies"
    elif "low" in v: risk_category = "low"
    elif "moderate" in v: risk_category = "moderate"
    elif "high" in v: risk_category = "high"
    else: risk_category = "unknown"
    return risk_category, risk_pct

df2[["risk_category", "risk_pct"]] = df2["risk_level"].apply(lambda x: pd.Series(clean_risk_level(x)))

print(df2["risk_category"].value_counts())
print("\nrisk_pct summary:\n", df2["risk_pct"].describe())
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.5</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Aggregate to disease level &amp; merge</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> We shrink the patient-by-patient data down to one summary row per disease, then glue the two datasets together using the cleaned-up disease names — keeping every row from both sides, even the ones that don't have a match yet.</p><p>Dataset 1 is collapsed from patient-level to one row per disease (symptom prevalence, average age, outcome-positivity rate), then outer-merged with dataset 2 on the canonical disease name so no row from either side is silently dropped.</p></div></div>

```python

# ============================================================
# STEP 6 — Aggregate df1 (patient-level) to one row per disease,
#          then outer-merge with df2 (already disease-level)
# ============================================================
agg_dict = {c: "mean" for c in symptom_cols}
agg_dict["age"] = "mean"

df1_agg = df1.groupby("disease_canonical").agg(agg_dict).reset_index()
df1_agg = df1_agg.rename(columns={c: f"{c}_prevalence" for c in symptom_cols})
df1_agg = df1_agg.rename(columns={"age": "avg_age"})

df1_agg["most_common_gender"] = df1_agg["disease_canonical"].map(
    df1.groupby("disease_canonical")["gender"].agg(lambda x: x.mode().iloc[0]))
df1_agg["outcome_positive_rate"] = df1_agg["disease_canonical"].map(
    df1.groupby("disease_canonical")["outcome_variable"].apply(lambda x: (x == "positive").mean()))
df1_agg["patient_count"] = df1_agg["disease_canonical"].map(df1.groupby("disease_canonical").size())

df2["symptom_list"] = df2["symptoms"].apply(
    lambda x: [s.strip() for s in str(x).split(",")] if pd.notna(x) else []
)
df2_dedup = df2.drop_duplicates(subset="disease_canonical", keep="first")

merged = pd.merge(
    df1_agg,
    df2_dedup[["disease_canonical", "symptoms", "symptom_list", "cures", "doctor", "risk_category", "risk_pct"]],
    on="disease_canonical", how="outer", indicator=True,
)
merged["source"] = merged["_merge"].map({"left_only": "df1_only", "right_only": "df2_only", "both": "both"})
merged = merged.drop(columns="_merge")

print(f"Merged shape: {merged.shape}")
print("\nSource breakdown:\n", merged["source"].value_counts())
merged.head()
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.6</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Final data-quality pass</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> One last check over the combined table to catch anything still missing or duplicated. Where we can fill in a gap honestly, we do — and we clearly mark that it was filled in rather than pretending it was always there.</p><p>Cleaning cannot manufacture more samples, but it <i>can</i> eliminate every genuine gap and be explicit about what's real vs. imputed:</p><ul style="padding-left:20px;margin:10px 0;"><li style="margin-bottom:5px;">Every row gets <b>some</b> symptom representation (text symptoms where available, else synthesized from the boolean flags).</li><li style="margin-bottom:5px;">Numeric prevalence fields missing entirely (df2-only rows have no patient stats) get the <b>global mean</b>, flagged with an <code>*_imputed</code> column — never silently faked.</li><li style="margin-bottom:5px;">Categorical/clinical fields (<code>cures</code>, <code>doctor</code>, <code>risk_category</code>) are labeled <code>"unknown"</code> rather than fabricated — inventing medical advice to fill a table is worse than leaving it missing.</li></ul></div></div>

```python

# ============================================================
# STEP 7 — Full data-quality pass: zero unexplained missingness, zero duplicates,
#          every gap explicitly flagged rather than silently imputed
# ============================================================
dup_check = merged["disease_canonical"].duplicated().sum()
assert dup_check == 0, "Deduplication failed — investigate before proceeding"

bool_symptom_names = {
    "fever_prevalence": "fever", "cough_prevalence": "cough",
    "fatigue_prevalence": "fatigue", "difficulty_breathing_prevalence": "difficulty breathing",
}

def synth_symptoms(row):
    if isinstance(row["symptom_list"], list) and len(row["symptom_list"]) > 0:
        return row["symptom_list"]
    present = [name for col, name in bool_symptom_names.items() if pd.notna(row[col]) and row[col] >= 0.5]
    return present  # may be [] if disease had none of the 4 tracked boolean symptoms

merged["symptom_list"] = merged.apply(synth_symptoms, axis=1)
merged["symptoms"] = merged["symptom_list"].apply(lambda x: ",".join(x) if x else "none reported")
merged["symptom_source"] = np.where(
    merged["cures"].notna(), "df2_text",
    np.where(merged["symptom_list"].map(len) > 0, "df1_flags_derived", "no_symptoms_flagged")
)

merged["has_patient_stats"] = merged["patient_count"].notna()
merged["has_clinical_detail"] = merged["cures"].notna()

numeric_impute_cols = ["fever_prevalence", "cough_prevalence", "fatigue_prevalence",
                        "difficulty_breathing_prevalence", "avg_age", "outcome_positive_rate"]
for c in numeric_impute_cols:
    merged[f"{c}_imputed"] = merged[c].isna()
    merged[c] = merged[c].fillna(merged[c].mean())

merged["patient_count"] = merged["patient_count"].fillna(0)
merged["most_common_gender"] = merged["most_common_gender"].fillna("unknown")
merged["risk_category"] = merged["risk_category"].fillna("unknown")
merged["doctor"] = merged["doctor"].fillna("unknown")
merged["cures"] = merged["cures"].fillna("unknown")
# risk_pct intentionally left NaN where genuinely unspecified ("varies") — no defensible impute

print("Post-cleaning completeness:\n", merged.isna().sum())
print(f"\nFinal shape: {merged.shape}")
print(f"Rows with full clinical + patient data (source == 'both'): {(merged['source']=='both').sum()}")

merged.to_csv("aggregated_disease_dataset_clean.csv", index=False)
print("\nSaved cleaned dataset -> aggregated_disease_dataset_clean.csv")
```

A normal machine-learning model needs many examples of something to learn to recognize it — with just one example, it literally cannot learn. So instead of forcing a model that would just be guessing, we use a smarter approach below.</p><p>The aggregated table has <b>one row per disease by construction</b>. If the modeling target is <i>"predict disease name"</i>, the number of classes always equals the number of rows — a classifier trained on this table can never see more than one example per class. No amount of cleaning fixes a sample-size problem; cleaning and modeling are separate concerns.</p><p>The chart below makes the ceiling concrete before we build anything on top of it.</p></div></div>

```python

# ============================================================
# Diagnostic: samples-per-class reality check
# ============================================================
class_sizes = merged["patient_count"].fillna(0)
fig, ax = plt.subplots(figsize=(9, 4.5))
ax.hist(class_sizes, bins=30, color=PALETTE["red"], alpha=0.8)
ax.set_title("Patient Records per Disease — the Small-N Problem")
ax.set_xlabel("Patient count for this disease"); ax.set_ylabel("Number of diseases")
plt.tight_layout()
plt.show()

single_sample = (class_sizes <= 1).sum()
print(f"Diseases with <= 1 patient record: {single_sample} out of {merged.shape[0]}")
print("A classifier trained to predict disease name from this table cannot generalize —")
print("this motivates the two complementary models built below instead of a single classifier.")
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.9</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Model 1a — Symptom-similarity matcher (primary prediction method)</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Rather than trying to force a single 'correct' answer, this compares a person's reported symptoms against every disease in the table and returns the closest matches — similar to how a doctor mentally runs through a list of possibilities rather than jumping to one diagnosis.</p><p>Instead of forcing a classifier onto a 155-class / ~1-sample-per-class problem, this builds a <b>cosine-similarity nearest-neighbor matcher</b> over symptom sets. It works even with a single example per disease and returns <b>ranked candidates</b> — behaving like a differential diagnosis rather than a single forced guess, which is both more useful and more honest given the data.</p></div></div>

```python

# ============================================================
# MODEL 1a — Build the unified symptom feature space
# ============================================================
def to_list(x):
    if isinstance(x, list):
        return x
    if isinstance(x, str):
        try:
            v = ast.literal_eval(x)
            return v if isinstance(v, list) else []
        except (ValueError, SyntaxError):
            return [s.strip() for s in x.split(",")] if x else []
    return []

df_model = merged.copy()
df_model["symptom_list"] = df_model["symptom_list"].apply(to_list)

def merge_bool_symptoms(row):
    s = set(row["symptom_list"])
    if row["fever_prevalence"] >= 0.5: s.add("fever")
    if row["cough_prevalence"] >= 0.5: s.add("cough")
    if row["fatigue_prevalence"] >= 0.5: s.add("fatigue")
    if row["difficulty_breathing_prevalence"] >= 0.5: s.add("difficulty breathing")
    return list(s)

df_model["symptom_set"] = df_model.apply(merge_bool_symptoms, axis=1)
df_model = df_model[df_model["symptom_set"].map(len) > 0].reset_index(drop=True)

mlb = MultiLabelBinarizer()
symptom_matrix = mlb.fit_transform(df_model["symptom_set"])

print(f"Diseases with usable symptom data: {df_model.shape[0]}")
print(f"Unique symptom vocabulary size: {len(mlb.classes_)}")
print(f"Feature matrix shape: {symptom_matrix.shape}")
```

```python

# ============================================================
# MODEL 1a — Symptom-similarity matcher, with a synonym-normalization layer
# so free-text user input ("shortness of breath") maps onto the dataset's
# vocabulary ("difficulty breathing") instead of silently missing a match
# ============================================================
symptom_synonyms = {
    "shortness of breath": "difficulty breathing", "sob": "difficulty breathing",
    "breathlessness": "difficulty breathing", "tiredness": "fatigue", "exhaustion": "fatigue",
}

def normalize_input_symptoms(symptom_list):
    return [symptom_synonyms.get(s.lower().strip(), s.lower().strip()) for s in symptom_list]

def predict_disease_from_symptoms(input_symptoms, top_n=5):
    # Rank candidate diseases by cosine similarity of symptom sets.
    normalized = normalize_input_symptoms(input_symptoms)
    input_vec = mlb.transform([normalized])
    sims = cosine_similarity(input_vec, symptom_matrix)[0]
    ranked_idx = sims.argsort()[::-1][:top_n]
    results = df_model.iloc[ranked_idx][["disease_canonical", "risk_category", "risk_pct", "cures", "doctor"]].copy()
    results["similarity_score"] = sims[ranked_idx]
    # Confidence score: normalize the returned candidates' similarity so they
    # sum to 100% -- expresses "how much more likely is this candidate than the
    # others shown", not an absolute probability (the underlying data can't
    # support a calibrated probability with ~1 sample/class).
    score_sum = results["similarity_score"].sum()
    results["confidence_pct"] = (results["similarity_score"] / score_sum * 100).round(1) if score_sum > 0 else 0.0
    return results.reset_index(drop=True)

print("=== Example: symptom-based disease prediction ===")
example_symptoms = ["fever", "cough", "fatigue", "shortness of breath"]
display(predict_disease_from_symptoms(example_symptoms))
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.9b</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Emergency case identification</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Some symptoms are serious enough that someone should get help right away, no matter which disease it eventually turns out to be. This step just checks for those obvious warning-sign combinations, like chest pain plus trouble breathing.</p><p>A symptom checker that only ranks candidate diseases misses an important safety behavior: some symptom <i>combinations</i> warrant urgent care regardless of which specific disease they turn out to be. This rule-based layer runs alongside (not instead of) the similarity matcher — a small, transparent red-flag list rather than a black-box severity model, since over- or under-triaging emergencies has real consequences and needs to be auditable.</p></div></div>

```python

# ============================================================
# Emergency case identification — red-flag symptom combinations
# ============================================================
EMERGENCY_SYMPTOM_FLAGS = {
    "chest pain", "difficulty breathing", "shortness of breath",
    "sudden weakness", "slurred speech", "loss of consciousness",
    "severe bleeding", "confusion",
}

def detect_emergency(input_symptoms, candidates=None):
    normalized = set(normalize_input_symptoms(input_symptoms))
    matched_flags = normalized & {normalize_input_symptoms([f])[0] for f in EMERGENCY_SYMPTOM_FLAGS}

    high_risk_candidate = False
    if candidates is not None and len(candidates):
        top = candidates.iloc[0]
        high_risk_candidate = (top.get("risk_category") == "high") or (pd.notna(top.get("risk_pct")) and top["risk_pct"] >= 20)

    # Two or more red-flag symptoms together, OR one red-flag symptom paired with
    # a high-risk top candidate, is treated as an emergency signal.
    is_emergency = len(matched_flags) >= 2 or (len(matched_flags) >= 1 and high_risk_candidate)

    return {
        "emergency_flag": is_emergency,
        "matched_red_flag_symptoms": sorted(matched_flags),
        "reason": "Multiple emergency-pattern symptoms reported" if len(matched_flags) >= 2
                   else "Emergency-pattern symptom combined with a high-risk candidate disease" if is_emergency
                   else "No emergency pattern detected",
    }

print("=== Example: emergency check ===")
candidates = predict_disease_from_symptoms(["chest pain", "shortness of breath", "sweating"])
print(detect_emergency(["chest pain", "shortness of breath", "sweating"], candidates))

print("\n=== Example: non-emergency check ===")
candidates2 = predict_disease_from_symptoms(["fatigue", "runny nose"])
print(detect_emergency(["fatigue", "runny nose"], candidates2))
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.10</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Model 1b — Patient-level outcome classifier (statistically sound target)</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Since we can't reliably predict the exact disease name (too few examples of each), we instead train a model for a simpler, more reliable question: given a patient's symptoms and profile, was their overall test result positive or negative? This has enough real examples to actually learn from.</p><p>Predicting <b>disease name</b> fails at 155 classes / ~1 sample each. But predicting <b><code>outcome_variable</code></b> (positive/negative) at the <b>patient level</b> (<code>df1</code>, 300 rows after cleaning) is a binary target with genuinely repeated labels — a fundamentally different, trainable problem. A first unregularized model showed 97.5% train vs 83% test accuracy (severe overfitting) with <code>age</code> as the top feature despite near-identical age distributions between outcome classes — a textbook sign of a tree memorizing noise. The model below is <b>regularized and validated with cross-validation</b>, and <code>age</code> is dropped after confirming it added no real signal.</p></div></div>

```python

# ============================================================
# MODEL 1b — Regularized patient-outcome classifier
# ============================================================
feature_cols = ["fever", "cough", "fatigue", "difficulty_breathing"]
cat_features = ["gender", "blood_pressure", "cholesterol_level"]

df_outcome_enc = pd.get_dummies(df1[feature_cols + cat_features], columns=cat_features)
X_outcome = df_outcome_enc.values
y_outcome = (df1["outcome_variable"] == "positive").astype(int).values

majority_baseline = max(Counter(y_outcome).values()) / len(y_outcome)
print(f"Class balance: {Counter(y_outcome)} | Majority baseline: {majority_baseline:.3f}")

clf_outcome = RandomForestClassifier(
    n_estimators=200, max_depth=4, min_samples_split=10, min_samples_leaf=5, random_state=42
)

cv = StratifiedKFold(n_splits=5, shuffle=True, random_state=42)
cv_scores = cross_val_score(clf_outcome, X_outcome, y_outcome, cv=cv, scoring="accuracy")
print(f"\n5-fold CV accuracy (regularized, no age): {cv_scores.mean():.3f} (+/- {cv_scores.std():.3f})")
print(f"Lift over majority baseline: {cv_scores.mean() - majority_baseline:+.3f}")

X_train, X_test, y_train, y_test = train_test_split(X_outcome, y_outcome, test_size=0.2, stratify=y_outcome, random_state=42)
clf_outcome.fit(X_train, y_train)
print(f"Train accuracy: {clf_outcome.score(X_train, y_train):.3f} | Test accuracy: {clf_outcome.score(X_test, y_test):.3f}")

y_pred_outcome = clf_outcome.predict(X_test)
print("\nClassification report (held-out test set):")
print(classification_report(y_test, y_pred_outcome, target_names=["negative", "positive"]))

feat_importance = pd.Series(clf_outcome.feature_importances_, index=df_outcome_enc.columns).sort_values(ascending=False)
fig, ax = plt.subplots(figsize=(8, 5))
feat_importance.plot(kind="barh", ax=ax, color=PALETTE["teal"])
ax.invert_yaxis()
ax.set_title("Outcome Classifier — Feature Importance (age excluded, no real signal found)")
plt.tight_layout()
plt.show()

# Fit the deployed model on all available rows
clf_outcome.fit(X_outcome, y_outcome)
print("\nFinal outcome classifier fit on all 300 rows and ready for use.")
```

<div style="background:#e8f8f5;border:1px solid #1abc9c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#1abc9c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">1.11</span><h3 style="margin:0;color:#0e8c76;font-size:1.14em;">Bridging functions — risk scoring &amp; treatment lookup (feeds Models 2 &amp; 3)</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Two small helper functions: one turns the list of possible diseases into one overall risk level, and the other looks up what treatment is on record for a given disease name.</p><p>Two small functions that consume the similarity matcher's output: one weights each candidate's risk category by its similarity score into a single risk label, the other looks up the recorded cures/doctor/risk for a named disease from the cleaned Kaggle table (Model 3 will supplement this with real-world medication data).</p></div></div>

```python

# ============================================================
# Risk assessment (weighted by symptom-match similarity) &
# treatment lookup (from the Kaggle "cures" column — supplemented later by Model 3)
# ============================================================
def assess_risk_from_symptoms(input_symptoms, top_n=3):
    matches = predict_disease_from_symptoms(input_symptoms, top_n=top_n)
    risk_weights = {"low": 1, "moderate": 2, "high": 3, "varies": 2, "unknown": 1}
    matches["risk_score"] = matches["risk_category"].map(risk_weights).fillna(1)
    weighted_risk = (matches["risk_score"] * matches["similarity_score"]).sum() / matches["similarity_score"].sum()
    emergency = detect_emergency(input_symptoms, matches)
    return {
        "top_candidates": matches[["disease_canonical", "similarity_score", "confidence_pct", "risk_category"]].to_dict("records"),
        "overall_risk_score": round(weighted_risk, 2),
        "overall_risk_label": "high" if weighted_risk >= 2.5 else "moderate" if weighted_risk >= 1.5 else "low",
        "emergency_flag": emergency["emergency_flag"],
        "emergency_reason": emergency["reason"],
    }

def recommend_treatment_from_table(disease_name):
    row = df_model[df_model["disease_canonical"] == disease_name]
    if row.empty:
        return {"error": f"'{disease_name}' not found in the aggregated Kaggle table"}
    row = row.iloc[0]
    return {
        "disease": disease_name, "cures": row["cures"],
        "recommended_doctor": row["doctor"], "risk_level": row["risk_category"],
        "risk_percentage": row["risk_pct"],
    }

print("=== Example: risk assessment ===")
print(assess_risk_from_symptoms(["fever", "cough", "shortness of breath", "chest pain"]))
print("\n=== Example: treatment lookup ===")
print(recommend_treatment_from_table("pneumonia"))
```

this</td></tr></tbody></table></div><p>BRFSS also <b>cannot be row-merged</b> with Models 1/3 — it has no disease-name key at all (a person can report multiple conditions simultaneously). It stands as its <b>own model</b>, called independently, exactly as scoped in the original project design.</p></div></div>

```python

# ============================================================
# STEP 1 — Download BRFSS 2015 and identify health-relevant columns
# ============================================================
_t0 = loading("Downloading CDC BRFSS 2015 (large file, may take a moment)...", "dl_model2")

brfss_path = kagglehub.dataset_download("cdc/behavioral-risk-factor-surveillance-system")
brfss = pd.read_csv(os.path.join(brfss_path, "2015.csv"), low_memory=False)

done("BRFSS 2015 downloaded &amp; loaded", "dl_model2", _t0)

print("Files found:", os.listdir(brfss_path))
print(f"\nFull shape: {brfss.shape}")

CONDITION_COLUMNS = {
    "diabetes": "DIABETE3", "heart_attack": "CVDINFR4", "coronary_hd": "CVDCRHD4",
    "stroke": "CVDSTRK3", "asthma": "ASTHMA3", "skin_cancer": "CHCSCNCR",
    "other_cancer": "CHCOCNCR", "arthritis": "HAVARTH3", "depression": "ADDEPEV2",
    "kidney_disease": "CHCKIDNY",
}
print("\nCondition columns present:", {k: v in brfss.columns for k, v in CONDITION_COLUMNS.items()})
```

<div style="background:#fef5e7;border:1px solid #f39c12;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#f39c12;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">2.2</span><h3 style="margin:0;color:#c9750d;font-size:1.14em;">BRFSS's coding scheme</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> This survey stores answers as number codes instead of plain words — for example, 1 means "yes" and 2 means "no". Before we can use any of it, we need to translate those codes back into something meaningful, and throw out the "don't know"/"refused" answers instead of treating them as real data.</p><p>BRFSS uses standard CDC survey coding across almost every field: <b>1 = Yes, 2 = No, 7 = Don't know, 9 = Refused</b>, with blanks for skip-logic. Treating 7/9 as a real third category would corrupt any model, so the decode step below explicitly drops don't-know/refused responses rather than encoding them as data.</p></div></div>

```python

# ============================================================
# STEP 2 — Decode helpers for BRFSS's coded fields
# ============================================================
def decode_yn(series):
    # 1=yes,2=no -> 1/0 ; 7/9/blank -> NaN (dropped, not a fake category)
    return series.map({1: 1, 2: 0}).where(series.isin([1, 2]))

def decode_alcdays(v):
    # ALCDAY5: 1xx = per week, 2xx = per month, 888 = none, 777/999 = DK/refused
    if pd.isna(v):
        return np.nan
    v = int(v)
    if v == 888: return 0
    if 101 <= v <= 199: return (v - 100) * 4.33     # per-week -> per-month estimate
    if 201 <= v <= 299: return v - 200
    return np.nan

RISK_FEATURE_COLS = ["_BMI5", "_AGEG5YR", "SEX", "_SMOKER3", "EXERANY2", "BPHIGH4", "TOLDHI2", "ALCDAY5"]

def prep_risk_features(raw):
    df = raw[RISK_FEATURE_COLS].copy()
    df["EXERANY2"] = decode_yn(df["EXERANY2"])
    df["TOLDHI2"] = decode_yn(df["TOLDHI2"])
    df["BPHIGH4"] = df["BPHIGH4"].map({1: 1, 2: 1, 3: 0, 4: 0}).where(df["BPHIGH4"].isin([1, 2, 3, 4]))
    df["_BMI5"] = df["_BMI5"].where(df["_BMI5"] < 9000) / 100    # BMI stored as BMI*100; drop sentinel values
    df["_AGEG5YR"] = df["_AGEG5YR"].where(df["_AGEG5YR"] <= 14)  # 5-year age-group code, 1..14 (80+)
    df["_SMOKER3"] = df["_SMOKER3"].where(df["_SMOKER3"] != 9)
    df["ALCDAY5"] = df["ALCDAY5"].apply(decode_alcdays)
    return df

risk_features = prep_risk_features(brfss)
print("Decoded feature preview:")
risk_features.head()
```

<div style="background:#fef5e7;border:1px solid #f39c12;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#f39c12;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">2.3</span><h3 style="margin:0;color:#c9750d;font-size:1.14em;">A Random Forest per condition — the multi-condition risk engine</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> We train nine separate, focused yes/no models — one just for diabetes, one just for heart attack, and so on — rather than one model trying to guess everything at once. Each one is also told to pay extra attention to catching real cases, since these conditions are rare in the general population.</p><p>Rather than one classifier for one disease, this trains <b>one binary classifier per BRFSS-tracked condition</b>, each using <code>class_weight="balanced"</code> since every one of these conditions is a small minority in the general population (a model that always predicts "no" would score high accuracy but catch zero true cases — the wrong failure mode for a screening tool). A progress bar tracks the 9-condition training loop.</p></div></div>

```python

# ============================================================
# STEP 3 — Train one risk classifier per condition, report ROC-AUC
# ============================================================
CONDITION_CODE_MAPS = {
    "diabetes": {1: 1, 2: 1, 3: 0, 4: 0},   # yes / yes-pregnancy-only counted positive; no / pre-diabetic negative
    "heart_attack": {1: 1, 2: 0}, "coronary_hd": {1: 1, 2: 0}, "stroke": {1: 1, 2: 0},
    "asthma": {1: 1, 2: 0}, "skin_cancer": {1: 1, 2: 0}, "other_cancer": {1: 1, 2: 0},
    "arthritis": {1: 1, 2: 0}, "depression": {1: 1, 2: 0}, "kidney_disease": {1: 1, 2: 0},
}

cat_cols = ["_AGEG5YR", "SEX", "_SMOKER3", "BPHIGH4"]
risk_models = {}

for name, condition_col in tqdm(list(CONDITION_COLUMNS.items()), desc="Training 9 BRFSS risk models"):
    code_map = CONDITION_CODE_MAPS[name]
    target = brfss[condition_col].map(code_map).where(brfss[condition_col].isin(code_map.keys()))

    combined = risk_features.copy()
    combined["target"] = target
    combined = combined.dropna()

    X = pd.get_dummies(combined.drop(columns="target"), columns=cat_cols)
    y = combined["target"].astype(int).values

    if len(Counter(y)) < 2 or min(Counter(y).values()) < 100:
        print(f"{name}: skipped (insufficient class balance)")
        continue

    X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, stratify=y, random_state=42)
    clf = RandomForestClassifier(n_estimators=50, max_depth=8, min_samples_leaf=20,
                                  class_weight="balanced", random_state=42, n_jobs=-1)
    clf.fit(X_train, y_train)
    y_pred = clf.predict(X_test)
    y_proba = clf.predict_proba(X_test)[:, 1]
    auc = roc_auc_score(y_test, y_proba)

    risk_models[name] = {
        "model": clf, "features": list(X.columns), "auc": auc,
        "n": len(y), "positive_rate": y.mean(),
        "precision": precision_score(y_test, y_pred, zero_division=0),
        "recall": recall_score(y_test, y_pred, zero_division=0),
        "f1": f1_score(y_test, y_pred, zero_division=0),
    }

print("\nPer-condition results:")
for name, info in risk_models.items():
    print(f"  {name:16s} n={info['n']:>7,}  positive_rate={info['positive_rate']:.3f}  ROC-AUC={info['auc']:.3f}")
```

<div style="background:#fef5e7;border-left:5px solid #f39c12;border-radius:10px;padding:16px 22px;font-family:'Segoe UI',Helvetica,Arial,sans-serif;box-shadow:0 2px 8px rgba(0,0,0,0.06);"><div style="font-weight:700;color:#c9750d;margin-bottom:6px;font-size:1.02em;">📐 Metric completeness note</div><div style="color:#37474f;font-size:0.95em;line-height:1.55;">Objectives.txt names precision/recall/F1 explicitly as required AI-model performance metrics. Reporting a bare ROC-AUC per condition isn't enough on its own — the summary table below gives the full picture at each model's default 0.5 decision threshold (before any recall-tuning, which is done separately for diabetes below).</div></div>

```python

# ============================================================
# Precision / Recall / F1 / ROC-AUC summary across all 9 condition models
# ============================================================
metrics_summary = pd.DataFrame({
    name: {"n": info["n"], "positive_rate": round(info["positive_rate"], 3),
           "precision": round(info["precision"], 3), "recall": round(info["recall"], 3),
           "f1": round(info["f1"], 3), "roc_auc": round(info["auc"], 3)}
    for name, info in risk_models.items()
}).T.sort_values("roc_auc", ascending=False)

print("Precision / Recall / F1 / ROC-AUC per condition (at default 0.5 threshold):")
display(metrics_summary)

# Visualize model quality across all conditions
auc_series = pd.Series({k: v["auc"] for k, v in risk_models.items()}).sort_values(ascending=True)
fig, ax = plt.subplots(figsize=(9, 5))
auc_series.plot(kind="barh", ax=ax, color=PALETTE["amber"])
ax.axvline(0.5, color="grey", linestyle="--", label="Random guessing")
ax.set_title("Risk Model Quality by Condition (ROC-AUC)")
ax.set_xlabel("ROC-AUC"); ax.legend()
plt.tight_layout()
plt.show()
```

<div style="background:#fef5e7;border:1px solid #f39c12;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#f39c12;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">2.4</span><h3 style="margin:0;color:#c9750d;font-size:1.14em;">Deep dive — diabetes threshold tuning for screening use</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> For diabetes specifically, we tune exactly how cautious the model should be. It's better to tell a healthy person to double-check with a doctor than to miss someone who's actually at risk, so we deliberately lean the model toward catching more real cases, even if that means a few extra false alarms.</p><p>Raw accuracy is misleading here: diabetes is a ~15% minority class, so a model that always predicts "no diabetes" already scores ~85% "accuracy" while catching zero true cases. For a <b>preliminary screening tool</b> (not a diagnostic one), the right trade-off favors <b>recall</b> — missing a true diabetic is worse than an extra false alarm, which just means "consider getting checked."</p></div></div>

```python

# ============================================================
# Precision/recall trade-off for the diabetes risk model, and a
# generalized callable risk-assessment function for any condition
# ============================================================
diabetes_model = risk_models["diabetes"]["model"]
diabetes_features = risk_models["diabetes"]["features"]

target = brfss["DIABETE3"].map(CONDITION_CODE_MAPS["diabetes"]).where(brfss["DIABETE3"].isin(CONDITION_CODE_MAPS["diabetes"].keys()))
combined = risk_features.copy(); combined["target"] = target
combined = combined.dropna()
X_d = pd.get_dummies(combined.drop(columns="target"), columns=cat_cols)
y_d = combined["target"].astype(int).values
_, X_test_d, _, y_test_d = train_test_split(X_d, y_d, test_size=0.2, stratify=y_d, random_state=42)

y_proba = diabetes_model.predict_proba(X_test_d)[:, 1]
precisions, recalls, thresholds = precision_recall_curve(y_test_d, y_proba)

fig, ax = plt.subplots(figsize=(7, 5))
ax.plot(recalls, precisions, color=PALETTE["red"], linewidth=2)
ax.set_xlabel("Recall"); ax.set_ylabel("Precision")
ax.set_title(f"Diabetes Risk Model — Precision/Recall Trade-off (ROC-AUC={risk_models['diabetes']['auc']:.3f})")
plt.tight_layout()
plt.show()

print("Recall / Precision at candidate operating points:")
for target_recall in [0.5, 0.6, 0.7, 0.8, 0.9]:
    idx = np.argmin(np.abs(recalls - target_recall))
    thr = thresholds[min(idx, len(thresholds) - 1)]
    print(f"  Recall~={recalls[idx]:.2f} -> Precision={precisions[idx]:.2f}, threshold={thr:.3f}")

CHOSEN_DIABETES_THRESHOLD = 0.53   # ~70% recall, screening-appropriate trade-off

def assess_condition_risk(condition, profile_row):
    # condition: one of risk_models keys (e.g. "diabetes")
    # profile_row: dict with keys matching RISK_FEATURE_COLS raw BRFSS codes
    info = risk_models[condition]
    row = pd.DataFrame([profile_row]).reindex(columns=info["features"], fill_value=0)
    proba = info["model"].predict_proba(row)[0, 1]
    threshold = CHOSEN_DIABETES_THRESHOLD if condition == "diabetes" else 0.5
    return {"condition": condition, "risk_probability": round(float(proba), 3),
            "flagged_at_risk": bool(proba >= threshold), "model_auc": round(info["auc"], 3)}

example_profile = pd.DataFrame([{
    "_BMI5": 31, "ALCDAY5": 8, "EXERANY2": 0, "TOLDHI2": 1,
}]).reindex(columns=diabetes_features, fill_value=0)
for code_val, col_prefix in [(6, "_AGEG5YR_"), (1, "SEX_"), (4, "_SMOKER3_"), (1, "BPHIGH4_")]:
    col = f"{col_prefix}{float(code_val)}"
    if col in example_profile.columns:
        example_profile[col] = 1

print("\nExample: 45yo male, BMI 31, no exercise, high BP, high cholesterol, moderate drinking")
proba = diabetes_model.predict_proba(example_profile)[0, 1]
print({"diabetes_risk_probability": round(float(proba), 3), "flagged_at_risk": bool(proba >= CHOSEN_DIABETES_THRESHOLD)})
```

This means:</p><ul style="padding-left:20px;margin:10px 0;"><li style="margin-bottom:5px;">Diagnosis and treatment must be <b>extracted from free text</b>, not read from a column.</li><li style="margin-bottom:5px;">Real diagnoses (<i>"Ascites from Portal HTN"</i>) rarely map cleanly onto the 155 clean disease names from Model 1 — they describe real clinical nuance, not tidy single labels.</li><li style="margin-bottom:5px;">Some notes have the diagnosis itself <b>redacted</b> (PII masking, shown as <code>___</code>) and must be filtered out rather than used as ground truth.</li><li style="margin-bottom:5px;">Medications here are <b>specific drugs + doses</b> (e.g. <i>"Furosemide 40 mg PO DAILY"</i>), a genuinely richer treatment signal than Model 1's generic <code>"antibiotics, rest, fluids"</code> style cures column.</li></ul><p>This is why Model 3 is built as its <b>own retrieval-based module</b>: a real-world treatment reference, not a replacement for Model 1's lookup table.</p><div style="background:#f7f9fb;border-left:5px solid #90a4ae;border-radius:10px;padding:16px 22px;font-family:'Segoe UI',Helvetica,Arial,sans-serif;box-shadow:0 2px 8px rgba(0,0,0,0.06);"><div style="font-weight:700;color:#455a64;margin-bottom:6px;font-size:1.02em;">⚠️ Deliberate deviation from objectives.txt's dataset-to-module table</div><div style="color:#37474f;font-size:0.95em;line-height:1.55;">That table marks MIMIC-IV as usable for Symptom Analysis and Disease Prediction as well as Risk Assessment and Treatment Recommendation. ```python

```python
# ============================================================
# STEP 2 — Confirm the notes are long-form narratives with a consistent
# section structure before attempting extraction
# ============================================================
text_col = "text" if "text" in mimic_df.columns else mimic_df.select_dtypes(include="object").columns[0]

word_counts = mimic_df[text_col].astype(str).apply(lambda x: len(x.split()))
print(f"Note length (words) -> mean: {word_counts.mean():.0f}, min: {word_counts.min()}, max: {word_counts.max()}")
print(f"Exact duplicate rows: {mimic_df.duplicated().sum()}")

common_headers = [
    "Chief Complaint", "History of Present Illness", "Past Medical History",
    "Discharge Diagnosis", "Discharge Medications", "Discharge Disposition",
    "Discharge Instructions", "Allergies", "Brief Hospital Course",
]

header_presence = {}
for header in common_headers:
    pct = mimic_df[text_col].astype(str).str.contains(re.escape(header), case=False, na=False).mean() * 100
    header_presence[header] = pct

fig, ax = plt.subplots(figsize=(9, 5))
pd.Series(header_presence).sort_values().plot(kind="barh", ax=ax, color=PALETTE["red"])
ax.set_title("Clinical Section-Header Coverage Across All Notes")
ax.set_xlabel("% of notes containing this section")
plt.tight_layout()
plt.show()
```

<div style="background:#fdedec;border:1px solid #e74c3c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#e74c3c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">3.3</span><h3 style="margin:0;color:#c0392b;font-size:1.14em;">Extraction pipeline — diagnosis + medications, with a redaction filter</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> We pull out just the diagnosis and medications from each note, and throw out any notes where that information was blacked out for patient privacy — so we only keep notes we can actually learn from.</p><p>Both <code>Discharge Diagnosis</code> and <code>Discharge Medications</code> appear in 96–98% of notes, so regex extraction is viable. Rows where the extracted diagnosis is dominated by redaction markers (<code>___</code>) are dropped — an honest measurement of unusable rows rather than silently keeping garbage labels. Progress bars track both extraction passes across all 999 notes.</p></div></div>

```python

# ============================================================
# STEP 3 — Regex extraction with a redaction filter
# ============================================================
def extract_section(text, section_name, next_sections):
    pattern = rf"{re.escape(section_name)}:?(.*?)(?:{'|'.join(re.escape(s) for s in next_sections)}|$)"
    match = re.search(pattern, str(text), re.IGNORECASE | re.DOTALL)
    return match.group(1).strip() if match else None

mimic_df["diagnosis_raw"] = mimic_df[text_col].progress_apply(
    lambda t: extract_section(t, "Discharge Diagnosis", ["Discharge Condition", "Discharge Instructions", "Discharge Disposition"])
)
mimic_df["medications_raw"] = mimic_df[text_col].progress_apply(
    lambda t: extract_section(t, "Discharge Medications", ["Discharge Disposition", "Discharge Condition", "Discharge Instructions"])
)

def redaction_ratio(text):
    if not text:
        return 1.0
    tokens = text.split()
    if not tokens:
        return 1.0
    return sum(1 for t in tokens if "___" in t) / len(tokens)

mimic_df["diagnosis_redaction_ratio"] = mimic_df["diagnosis_raw"].apply(redaction_ratio)

before = mimic_df.shape[0]
mimic_clean = mimic_df[
    mimic_df["diagnosis_raw"].notna()
    & mimic_df["medications_raw"].notna()
    & (mimic_df["diagnosis_redaction_ratio"] < 0.4)   # drop notes where the diagnosis is mostly redacted
].copy()
mimic_clean["diagnosis_clean"] = mimic_clean["diagnosis_raw"].str.replace(r"\s+", " ", regex=True).str.strip()
mimic_clean["medications_clean"] = mimic_clean["medications_raw"].str.replace(r"\s+", " ", regex=True).str.strip()

print(f"Notes with both sections extracted, before redaction filter: {mimic_df['diagnosis_raw'].notna().sum() & mimic_df['medications_raw'].notna().sum()}")
print(f"Notes usable after redaction filter: {mimic_clean.shape[0]} / {before} ({mimic_clean.shape[0]/before*100:.1f}%)")

mimic_clean[["diagnosis_clean", "medications_clean"]].head(5)
```

<div style="background:#fdedec;border:1px solid #e74c3c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#e74c3c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">3.4</span><h3 style="margin:0;color:#c0392b;font-size:1.14em;">The treatment-recommendation model — TF-IDF retrieval</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> Instead of needing an exact word-for-word match, this compares the meaning of a new diagnosis to real past notes and pulls up the medications those similar patients actually received — so it still works even if the wording is a little different.</p><p>Rather than a flat lookup table (fragile — an exact-string diagnosis match rarely occurs twice), this builds a <b>TF-IDF vector space over all extracted diagnoses</b> and retrieves the most similar historical notes for any new diagnosis description via cosine similarity — a genuine content-based retrieval model that generalizes to diagnosis phrasings not seen verbatim before.</p></div></div>

```python

# ============================================================
# MODEL 3 — TF-IDF diagnosis-similarity retrieval model
# ============================================================
tfidf = TfidfVectorizer(stop_words="english", max_features=2000, ngram_range=(1, 2))
diagnosis_matrix = tfidf.fit_transform(mimic_clean["diagnosis_clean"])

def recommend_treatment_from_diagnosis(query_text, top_n=3):
    # Given a free-text diagnosis description, retrieve the most similar real
    # discharge notes and return their actual prescribed medications.
    query_vec = tfidf.transform([query_text])
    sims = cosine_similarity(query_vec, diagnosis_matrix)[0]
    ranked_idx = sims.argsort()[::-1][:top_n]
    results = mimic_clean.iloc[ranked_idx][["diagnosis_clean", "medications_clean"]].copy()
    results["similarity_score"] = sims[ranked_idx]
    return results.reset_index(drop=True)

print("=== Example: retrieval for a new diagnosis description ===")
display(recommend_treatment_from_diagnosis("congestive heart failure with shortness of breath", top_n=3))

print("\n=== Example: retrieval for pneumonia-like diagnosis ===")
display(recommend_treatment_from_diagnosis("pneumonia with fever and cough", top_n=3))
```

<div style="background:#fdedec;border:1px solid #e74c3c;border-radius:12px;padding:20px 24px;margin-top:16px;box-shadow:0 3px 10px rgba(0,0,0,0.07);font-family:'Segoe UI',Helvetica,Arial,sans-serif;"><div style="display:flex;align-items:center;gap:12px;margin-bottom:12px;"><span style="background:#e74c3c;color:white;font-weight:700;font-size:0.78em;padding:5px 11px;border-radius:6px;min-width:40px;text-align:center;box-shadow:0 2px 6px rgba(0,0,0,0.15);flex-shrink:0;">3.5</span><h3 style="margin:0;color:#c0392b;font-size:1.14em;">Quantitative evaluation — "recommendation relevance"</h3></div><div style="color:#37474f;font-size:0.95em;line-height:1.65;"><p style="background:rgba(255,255,255,0.65);border-radius:8px;padding:10px 14px;margin:0 0 12px 0;color:#37474f;"><b>💬 In plain terms:</b> We test how good these recommendations really are: hide some real notes from the model, ask it to guess their treatment from the diagnosis alone, then check how close its guess is to what the patient was actually given — compared against just guessing randomly.</p><p>The two examples above are qualitative. Objectives.txt names <b>recommendation relevance</b> as a Healthcare Performance metric, so this section measures it: hold out 20% of notes, build the TF-IDF space on the remaining 80% only, then for each held-out note check whether the medications of its nearest retrieved training note(s) actually overlap with what that patient was really prescribed (token-level Jaccard overlap between medication lists, since exact-string drug/dose matches are rare but overlapping drug names are meaningful). This is compared against a random-retrieval baseline so the score is interpretable.</p></div></div>

```python

# ============================================================
# MODEL 3 — Held-out evaluation of retrieval relevance
# ============================================================
def med_tokens(text):
    return {w for w in re.findall(r"[a-zA-Z]{4,}", str(text).lower())}

mimic_train, mimic_test = train_test_split(mimic_clean, test_size=0.2, random_state=42)

tfidf_eval = TfidfVectorizer(stop_words="english", max_features=2000, ngram_range=(1, 2))
train_matrix = tfidf_eval.fit_transform(mimic_train["diagnosis_clean"])
train_med_tokens = mimic_train["medications_clean"].apply(med_tokens).values

rng = np.random.RandomState(42)
top1_scores, top3_scores, random_baseline_scores = [], [], []

def jaccard(a, b):
    union = a | b
    return len(a & b) / len(union) if union else 0.0

for _, row in tqdm(mimic_test.iterrows(), total=len(mimic_test), desc="Evaluating retrieval relevance"):
    true_tokens = med_tokens(row["medications_clean"])
    if not true_tokens:
        continue

    query_vec = tfidf_eval.transform([row["diagnosis_clean"]])
    sims = cosine_similarity(query_vec, train_matrix)[0]
    ranked_idx = sims.argsort()[::-1][:3]

    top1_scores.append(jaccard(true_tokens, train_med_tokens[ranked_idx[0]]))
    top3_scores.append(max(jaccard(true_tokens, train_med_tokens[i]) for i in ranked_idx))

    random_idx = rng.randint(0, len(mimic_train))
    random_baseline_scores.append(jaccard(true_tokens, train_med_tokens[random_idx]))

print(f"Held-out notes evaluated: {len(top1_scores)}")
print(f"Mean top-1 medication-token relevance:  {np.mean(top1_scores):.3f}")
print(f"Mean best-of-top-3 relevance:            {np.mean(top3_scores):.3f}")
print(f"Random-retrieval baseline relevance:     {np.mean(random_baseline_scores):.3f}")
print(f"Lift over random baseline (top-1):       {np.mean(top1_scores) - np.mean(random_baseline_scores):+.3f}")
```

```python

# ============================================================
# UNIFIED PIPELINE — orchestrates all three models behind one call,
# without merging their underlying data
# ============================================================
def medassist_pipeline(symptoms, lifestyle_profile=None, risk_condition="diabetes"):
    # symptoms: list[str] of user-reported symptoms
    # lifestyle_profile: dict matching risk_models[risk_condition]["features"] raw inputs (optional)
    # risk_condition: which BRFSS condition to screen for, if a profile is given
    report = {}

    # Model 1 — symptom -> ranked disease candidates
    candidates = predict_disease_from_symptoms(symptoms, top_n=3)
    report["disease_candidates"] = candidates.to_dict("records")
    report["risk_assessment_from_symptoms"] = assess_risk_from_symptoms(symptoms, top_n=3)

    # Model 2 — optional population-level risk screening
    if lifestyle_profile is not None:
        report["population_risk_screening"] = assess_condition_risk(risk_condition, lifestyle_profile)

    # Model 3 — treatment reference for the top candidate disease
    top_disease = candidates.iloc[0]["disease_canonical"] if len(candidates) else None
    if top_disease:
        report["treatment_from_kaggle_table"] = recommend_treatment_from_table(top_disease)
        report["treatment_from_real_notes"] = recommend_treatment_from_diagnosis(top_disease, top_n=2).to_dict("records")

    return report

# Demo: full multimodal report for a single user query
demo_report = medassist_pipeline(
    symptoms=["fever", "cough", "fatigue", "shortness of breath"],
    lifestyle_profile={"_BMI5": 31, "ALCDAY5": 8, "EXERANY2": 0, "TOLDHI2": 1,
                        "_AGEG5YR_6.0": 1, "SEX_1.0": 1, "_SMOKER3_4.0": 1, "BPHIGH4_1.0": 1},
    risk_condition="diabetes",
)

import pprint
pprint.pprint(demo_report)
```
